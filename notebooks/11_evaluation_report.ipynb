### Known Limitations
* **Historical Data Scope**: The dataset (Kim et al., WWW '20) covers 2012-02-12 to 2019-05-15. "Current" predictions are based on patterns up to 2019-05-15. It does not include newer features like Instagram Reels.
* **Timestamps**: All posting times are analyzed in UTC because audience geography is unknown (mixture). Conversions are provided based on the input timezone using standard offset logic.
* **Followers Constraint**: Follower counts were captured at the time of scrape, which represents the account state at the end of the data period, rather than exactly at the time of posting.
* **Diagnostic Limitations**: The Category mapping (e.g. mapping "Family" test brands to "Interior") may have inaccuracies due to differences between the longer synthetic profiles and typical short Instagram captions.

## 3. Backend Handover Guide

### Artifacts Directory (`artifacts/v1.0/`)
The backend service must have access to the following artifacts during inference:
- `trends.parquet`
- `content_type.parquet`
- `historical_engagement.parquet`
- `posting_time.parquet`
- `category_centroids.pkl`
- `hashtag_embeddings.parquet`
- `fallback_medians.json`
- `model.joblib`
- `label_thresholds.json`

### Pipeline Usage
Instantiate the pipeline once at service startup to load all artifacts into memory:
```python
from pipeline import RecommenderPipeline
pipeline = RecommenderPipeline()
```

For each request, call the `recommend()` method:
```python
result_dict = pipeline.recommend(input_dict)
```

### Contract Schema
Inputs must conform to `contract/input_schema.json`.
Outputs are guaranteed to conform to `contract/output_schema.json`.

**Required Inputs:**
- `brand_name` (string)
- `business_description` (string)
- `industry` (string)
- `target_audience` (string)

**Optional Additions (must be supported by the Backend):**
- `follower_count` (integer) - Highly recommended for better engagement predictions.
- `timezone` (string) - IANA format, defaults to UTC.
- `brand_tone` (string) - Passed through/ignored by ML, but used by the LLM.

**What NOT to recompute:**
The ML output is complete. The Backend must NOT recalculate relevance, trend direction, predictions, or timezone conversions.
